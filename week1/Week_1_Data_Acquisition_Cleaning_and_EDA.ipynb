```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("fake_job_postings.csv")

print("Dataset loaded successfully!")
```

```python
print("Shape of the dataset means total rows and columns:")
print("Shape:", df.shape)
```

```python
print("number of rows :", df.shape[0])
print("number of columns:",df.shape[1])
```

```python
print(" view the first 5 rows:")
df.head()
```

```python
print("view the last 5 rows:")
df.tail()
```

```python
print("columns in the dataset:")
x =df.columns
print(list(x))
```

```python
print("Checking the data types and non-null values:")
df.info()
```

```python
missing = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": df.isnull().mean() * 100
})

missing = missing.sort_values(
    by="Missing Percentage",
    ascending=False
)

missing
```

```python
print("Visualize missing values: ")
plt.figure(figsize=(12, 6))

missing["Missing Percentage"].plot(kind="bar")

plt.title("Missing Values by Column")
plt.xlabel("Columns")
plt.ylabel("Missing Percentage (%)")
plt.xticks(rotation=90)

plt.tight_layout()
plt.show()
```

```python
print("Check Duplicate Rows:")
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
if duplicate_count > 0:
    df = df.drop_duplicates()
    print("Duplicates removed.")
else:
    print("No duplicate rows found.")
```

```python
print("Dataset shape after duplicate removal:", df.shape)
```

```python
print("Checking the Target Variable:")
```

```python
print(df["fraudulent"].value_counts())
```

```python
print(df["fraudulent"].value_counts(normalize=True) * 100)
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="fraudulent")

plt.title("Genuine vs Fraudulent Job Postings")
plt.xlabel("Job Type")
plt.ylabel("Number of Job Postings")
plt.xticks([0, 1], ["Genuine", "Fraudulent"])

plt.tight_layout()
plt.show()
```

```python
print(df.dtypes)
```

```python
binary_columns = [
    "telecommuting",
    "has_company_logo",
    "has_questions",
    "fraudulent"
]

for col in binary_columns:
    print("\n", col)
    print(df[col].value_counts(dropna=False))
```

```python

text_columns = [
    "title",
    "location",
    "department",
    "salary_range",
    "company_profile",
    "description",
    "requirements",
    "benefits",
    "employment_type",
    "required_experience",
    "required_education",
    "industry",
    "function"
]

for col in text_columns:
    if col in df.columns:
        df[col] = df[col].fillna("")
```

```python
print(df.isnull().sum())
```

```python
print("Create new feature:")
df["description_length"] = (
    df["description"]
    .astype(str)
    .str.len()
)
```

```python
df[["description", "description_length"]].head()
```

```python
df["description_length"].describe()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="description_length",
    bins=50
)

plt.title("Distribution of Job Description Length")
plt.xlabel("Description Length")
plt.ylabel("Number of Job Postings")

plt.tight_layout()
plt.show()
```

```python
print("Analyze Employment Type:")
df["employment_type"].value_counts()
```

```python
plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    y="employment_type",
    order=df["employment_type"].value_counts().index
)

plt.title("Distribution of Employment Types")
plt.xlabel("Number of Job Postings")
plt.ylabel("Employment Type")

plt.tight_layout()
plt.show()
```

```python
print("Analyze Company Logo vs Fraud")
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="has_company_logo",
    hue="fraudulent"
)

plt.title("Company Logo Presence vs Fraudulent Job Postings")
plt.xlabel("Has Company Logo")
plt.ylabel("Number of Job Postings")

plt.tight_layout()
plt.show()
```

```python
print("Analyze Telecommuting vs Fraud")
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="telecommuting",
    hue="fraudulent"
)

plt.title("Telecommuting vs Fraudulent Job Postings")
plt.xlabel("Telecommuting")
plt.ylabel("Number of Job Postings")

plt.tight_layout()
plt.show()
```

```python
print("Save the Cleaned Dataset : ")
df.to_csv(
    "cleaned_fake_job_postings.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
